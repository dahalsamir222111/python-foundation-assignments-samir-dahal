```python
"""
Exercise: File Validator
Student: [Samir Dahal]
Day: 1
"""

# Ask user to enter a file name
file_name = input("Enter the file name: ")

# Clean the input - remove spaces and convert to lowercase
file_name = file_name.strip().lower()

# Check if the file is valid
if file_name.endswith(".csv"):
    print(f" {file_name} is a valid CSV file.")
elif file_name.endswith(".json"):
    print(f" {file_name} is a valid JSON file.")
elif file_name.endswith(".parquet"):
    print(f" {file_name} is a valid PARQUET file.")
else:
    print(f" {file_name} is not a valid format. Please use .csv, .json, or .parquet")

print("\n--- Testing with sample values ---")

# Test with sample values
test_files = ["sales.csv", "CUSTOMERS.JSON", "transactions.parquet", "report.xlsx"]

for test_file in test_files:
    # Clean each test file
    test_file_clean = test_file.strip().lower()
    
    # Check each test file
    if test_file_clean.endswith(".csv"):
        print(f" {test_file} is valid (CSV)")
    elif test_file_clean.endswith(".json"):
        print(f" {test_file} is valid (JSON)")
    elif test_file_clean.endswith(".parquet"):
        print(f" {test_file} is valid (PARQUET)")
    else:
        print(f" {test_file} is NOT valid")
```

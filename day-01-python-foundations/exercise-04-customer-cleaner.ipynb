```python
"""
Exercise: Customer Record Cleaner
Student: Samir Dahal
Day: 1
"""

# Raw data (messy input)
raw_name = "  sAMiR DaHal "
raw_city = "kATHMANDU "
raw_age = "21"
raw_email = " SAMIR@MAIL.COM "

# Clean the values
clean_name = raw_name.strip().title()        # Remove spaces and convert to Title Case
clean_city = raw_city.strip().capitalize()   # Remove spaces and capitalize first letter
clean_age = int(raw_age.strip())             # Remove spaces and convert to integer
clean_email = raw_email.strip().lower()      # Remove spaces and convert to lowercase

# Determine status using ternary expression
status = "Adult" if clean_age >= 18 else "Minor"

# Display output
print(f"Name: {clean_name}")
print(f"City: {clean_city}")
print(f"Age: {clean_age}")
print(f"Email: {clean_email}")
print(f"Status: {status}")
```

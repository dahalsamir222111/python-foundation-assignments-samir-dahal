```python
"""
Exercise: Pipeline Health Status
Student: Samir Dahal
Day: 1
"""

# Test Case 1
print("=" * 50)
print("TEST CASE 1")
print("=" * 50)

rows_loaded = 9800
rows_failed = 200
runtime_minutes = 18

# Calculate failure rate
total_rows = rows_loaded + rows_failed
failure_rate = (rows_failed / total_rows) * 100

# Determine status
if failure_rate <= 2 and runtime_minutes <= 20:
    status = "Healthy"
elif failure_rate <= 5:
    status = "Warning"
else:
    status = "Critical"

print(f"Rows Loaded: {rows_loaded}")
print(f"Rows Failed: {rows_failed}")
print(f"Total Rows: {total_rows}")
print(f"Failure Rate: {failure_rate:.2f}%")
print(f"Runtime: {runtime_minutes} minutes")
print(f"Pipeline Status: {status}")

# Test Case 2
print("\n" + "=" * 50)
print("TEST CASE 2")
print("=" * 50)

rows_loaded = 9500
rows_failed = 500
runtime_minutes = 15

total_rows = rows_loaded + rows_failed
failure_rate = (rows_failed / total_rows) * 100

if failure_rate <= 2 and runtime_minutes <= 20:
    status = "Healthy"
elif failure_rate <= 5:
    status = "Warning"
else:
    status = "Critical"

print(f"Rows Loaded: {rows_loaded}")
print(f"Rows Failed: {rows_failed}")
print(f"Total Rows: {total_rows}")
print(f"Failure Rate: {failure_rate:.2f}%")
print(f"Runtime: {runtime_minutes} minutes")
print(f"Pipeline Status: {status}")

# Test Case 3 - Important: Low failure rate but high runtime
print("\n" + "=" * 50)
print("TEST CASE 3")
print("=" * 50)

rows_loaded = 9900
rows_failed = 100
runtime_minutes = 30

total_rows = rows_loaded + rows_failed
failure_rate = (rows_failed / total_rows) * 100

if failure_rate <= 2 and runtime_minutes <= 20:
    status = "Healthy"
elif failure_rate <= 5:
    status = "Warning"
else:
    status = "Critical"

print(f"Rows Loaded: {rows_loaded}")
print(f"Rows Failed: {rows_failed}")
print(f"Total Rows: {total_rows}")
print(f"Failure Rate: {failure_rate:.2f}%")
print(f"Runtime: {runtime_minutes} minutes")
print(f"Pipeline Status: {status}")

print("\n" + "=" * 50)
print("CONCLUSION")
print("=" * 50)
print("Test Case 1: Failure rate is 2% and runtime is 18 min -> HEALTHY")
print("Test Case 2: Failure rate is 5% and runtime is 15 min -> WARNING")
print("Test Case 3: Failure rate is 1% but runtime is 30 min -> CRITICAL")
print("\n Important: Even though failure rate is low (1%),")
print("   the pipeline is CRITICAL because runtime exceeds 20 minutes.")
```

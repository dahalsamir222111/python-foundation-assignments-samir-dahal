```python
"""
Exercise: Clean Numeric Values
Student: Samir Dahal
Day: 2
"""

# Given raw values
raw_values = [100, None, 250, "invalid", 300, None, 450]

# --- Method 1: Using a loop, continue, and isinstance() ---
print("Method 1: Using Loop")
clean_values_loop = []

for value in raw_values:
    # Check if value is an integer
    if isinstance(value, int):
        clean_values_loop.append(value)
    else:
        continue  # Skip non-integer values

print(f"Cleaned values (loop): {clean_values_loop}")

# --- Method 2: Using list comprehension ---
print("\nMethod 2: Using List Comprehension")
clean_values_comprehension = [value for value in raw_values if isinstance(value, int)]

print(f"Cleaned values (comprehension): {clean_values_comprehension}")

```

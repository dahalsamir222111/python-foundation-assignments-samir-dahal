```python
"""
Exercise: Sales List Analysis
Student: Samir Dahal
Day: 2
"""

# Given monthly sales data
monthly_sales = [85000, 120000, 95000, 140000, 75000, 160000]

print("=" * 50)
print("SALES LIST ANALYSIS")
print("=" * 50)

# 1. Sorted list from highest to lowest
sorted_sales = sorted(monthly_sales, reverse=True)
print(f"\n1. Sales sorted (highest to lowest):")
print(f"   {sorted_sales}")

# 2. List containing only values above 100000
high_sales = [amount for amount in monthly_sales if amount > 100000]
print(f"\n2. Sales above 100,000:")
print(f"   {high_sales}")

# 3. List where each amount has 13% tax added
tax_rate = 0.13
sales_with_tax = [amount + (amount * tax_rate) for amount in monthly_sales]
print(f"\n3. Sales with 13% tax added:")
print(f"   {sales_with_tax}")

# 4. Total sales amount
total_sales = sum(monthly_sales)
print(f"\n4. Total sales amount:")
print(f"   NPR {total_sales:,.2f}")

# 5. Average sales amount
average_sales = sum(monthly_sales) / len(monthly_sales)
print(f"\n5. Average sales amount:")
print(f"   NPR {average_sales:,.2f}")

print("\n" + "=" * 50)
print("SUMMARY")
print("=" * 50)
print(f"Highest sale: NPR {max(monthly_sales):,}")
print(f"Lowest sale: NPR {min(monthly_sales):,}")
print(f"Number of months: {len(monthly_sales)}")
```

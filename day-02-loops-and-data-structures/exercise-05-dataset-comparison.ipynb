```python
"""
Exercise: Dataset Comparison
Student: Samir Dahal
Day: 2
"""

# Given datasets
dataset_a = {
    "customer",
    "sales",
    "product",
    "employee"
}

dataset_b = {
    "sales",
    "product",
    "supplier",
    "inventory"
}

print("=" * 50)
print("DATASET COMPARISON")
print("=" * 50)

# 1. All unique dataset names (union)
all_datasets = dataset_a | dataset_b  # or dataset_a.union(dataset_b)
print(f"\n1. All unique dataset names:")
print(f"   {all_datasets}")

# 2. Datasets found in both groups (intersection)
both_datasets = dataset_a & dataset_b  # or dataset_a.intersection(dataset_b)
print(f"\n2. Datasets found in both groups:")
print(f"   {both_datasets}")

# 3. Datasets only in dataset_a (difference)
only_a = dataset_a - dataset_b  # or dataset_a.difference(dataset_b)
print(f"\n3. Datasets only in dataset_a:")
print(f"   {only_a}")

# 4. Datasets only in dataset_b (difference)
only_b = dataset_b - dataset_a  # or dataset_b.difference(dataset_a)
print(f"\n4. Datasets only in dataset_b:")
print(f"   {only_b}")

print("\n" + "=" * 50)
print("SUMMARY")
print("=" * 50)

# Additional info
print(f"Total unique datasets: {len(all_datasets)}")
print(f"Datasets in dataset_a: {len(dataset_a)}")
print(f"Datasets in dataset_b: {len(dataset_b)}")
print(f"Common datasets: {len(both_datasets)}")
print(f"Only in dataset_a: {len(only_a)}")
print(f"Only in dataset_b: {len(only_b)}")
```

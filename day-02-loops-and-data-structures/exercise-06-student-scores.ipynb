```python
"""
Exercise: Student Score Dictionary
Student: Samir Dahal
Day: 2
"""

# Given student scores
student_scores = {
    "Anisha": 78,
    "Ravi": 55,
    "Maya": 92,
    "Sagar": 61,
    "Nima": 48
}

print("=" * 50)
print("STUDENT SCORE DICTIONARY")
print("=" * 50)

# Task 1: Print every student and score
print("\n1. All students and their scores:")
for name, score in student_scores.items():
    print(f"   {name}: {score}")

# Task 2: Dictionary with only students who scored at least 60
passing_students = {name: score for name, score in student_scores.items() if score >= 60}
print("\n2. Students who scored 60 or above:")
for name, score in passing_students.items():
    print(f"   {name}: {score}")

# Task 3: Find the student with the highest score
highest_score = max(student_scores.values())
highest_student = [name for name, score in student_scores.items() if score == highest_score]
print(f"\n3. Highest score: {highest_score} (Student: {highest_student[0]})")

# Task 4: Calculate the average score
average_score = sum(student_scores.values()) / len(student_scores)
print(f"\n4. Average score: {average_score:.2f}")

print("\n" + "=" * 50)
print("SUMMARY")
print("=" * 50)

# Additional analysis
print(f"Total students: {len(student_scores)}")
print(f"Passing students (≥60): {len(passing_students)}")
failing_students = len(student_scores) - len(passing_students)
print(f"Failing students (<60): {failing_students}")

# Find lowest score
lowest_score = min(student_scores.values())
lowest_student = [name for name, score in student_scores.items() if score == lowest_score]
print(f"Lowest score: {lowest_score} (Student: {lowest_student[0]})")
```

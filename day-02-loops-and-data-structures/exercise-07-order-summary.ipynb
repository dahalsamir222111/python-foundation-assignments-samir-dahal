```python
"""
Exercise: Nested Order Summary
Student: Samir Dahal
Day: 2
"""

# Given nested orders dictionary
orders = {
    "ORD-001": {
        "customer": "Anisha",
        "amount": 2500,
        "status": "Completed"
    },
    "ORD-002": {
        "customer": "Ravi",
        "amount": 1800,
        "status": "Pending"
    },
    "ORD-003": {
        "customer": "Maya",
        "amount": 3200,
        "status": "Completed"
    }
}

print("=" * 50)
print("NESTED ORDER SUMMARY")
print("=" * 50)

# Task 1: Print every order ID and customer
print("\n1. All Orders and Customers:")
for order_id, order_details in orders.items():
    print(f"   {order_id}: {order_details['customer']}")

# Task 2: Print only completed orders
print("\n2. Completed Orders:")
for order_id, order_details in orders.items():
    if order_details['status'] == "Completed":
        print(f"   {order_id}: {order_details['customer']} - NPR {order_details['amount']}")

# Task 3: Calculate total amount of completed orders
completed_total = 0
for order_id, order_details in orders.items():
    if order_details['status'] == "Completed":
        completed_total += order_details['amount']
print(f"\n3. Total amount of completed orders: NPR {completed_total}")

# Task 4: Count pending orders
pending_count = 0
for order_id, order_details in orders.items():
    if order_details['status'] == "Pending":
        pending_count += 1
print(f"\n4. Pending orders: {pending_count}")

# Task 5: Add a new order
orders["ORD-004"] = {
    "customer": "Samir",
    "amount": 1500,
    "status": "Pending"
}
print(f"\n5. Added new order ORD-004 for Samir (NPR 1500, Pending)")

# Display updated orders
print("\n" + "=" * 50)
print("UPDATED ORDERS")
print("=" * 50)

# Show all orders with details
print("\nAll Orders:")
for order_id, order_details in orders.items():
    print(f"\n   {order_id}:")
    print(f"      Customer: {order_details['customer']}")
    print(f"      Amount: NPR {order_details['amount']}")
    print(f"      Status: {order_details['status']}")

print("\n" + "=" * 50)
print("SUMMARY")
print("=" * 50)

# Additional analysis
total_orders = len(orders)
completed_orders = 0
pending_orders = 0
total_amount = 0

for order_id, order_details in orders.items():
    total_amount += order_details['amount']
    if order_details['status'] == "Completed":
        completed_orders += 1
    elif order_details['status'] == "Pending":
        pending_orders += 1

print(f"Total orders: {total_orders}")
print(f"Completed: {completed_orders}")
print(f"Pending: {pending_orders}")
print(f"Total order amount: NPR {total_amount}")
```

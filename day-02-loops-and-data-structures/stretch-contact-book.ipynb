```python
"""
Exercise: Contact Book Menu (Stretch)
Student: Samir Dahal
Day: 2
"""

# Initialize empty contact book
contact_book = {}

print("=" * 50)
print("WELCOME TO CONTACT BOOK")
print("=" * 50)

while True:
    # Display menu
    print("\n--- MENU ---")
    print("1. Add contact")
    print("2. Search contact")
    print("3. Delete contact")
    print("4. Display all contacts")
    print("5. Exit")
    
    # Get user choice
    choice = input("\nEnter your choice (1-5): ")
    
    # Option 1: Add contact
    if choice == "1":
        print("\n--- Add New Contact ---")
        name = input("Enter name: ").strip().title()
        
        # Check if contact already exists
        if name in contact_book:
            print(f" Contact '{name}' already exists!")
            overwrite = input("Do you want to overwrite? (yes/no): ").lower()
            if overwrite != "yes":
                print("Contact not added.")
                continue
        
        # Get contact details
        phone = input("Enter phone number: ")
        email = input("Enter email address: ").strip().lower()
        
        # Add to contact book
        contact_book[name] = {
            "phone": phone,
            "email": email
        }
        print(f" Contact '{name}' added successfully!")
    
    # Option 2: Search contact
    elif choice == "2":
        print("\n--- Search Contact ---")
        name = input("Enter name to search: ").strip().title()
        
        if name in contact_book:
            print(f"\n Contact: {name}")
            print(f"   Phone: {contact_book[name]['phone']}")
            print(f"   Email: {contact_book[name]['email']}")
        else:
            print(f" Contact '{name}' not found.")
    
    # Option 3: Delete contact
    elif choice == "3":
        print("\n--- Delete Contact ---")
        name = input("Enter name to delete: ").strip().title()
        
        if name in contact_book:
            confirm = input(f"Are you sure you want to delete '{name}'? (yes/no): ").lower()
            if confirm == "yes":
                del contact_book[name]
                print(f" Contact '{name}' deleted successfully!")
            else:
                print("Deletion cancelled.")
        else:
            print(f" Contact '{name}' not found.")
    
    # Option 4: Display all contacts
    elif choice == "4":
        print("\n--- All Contacts ---")
        
        if not contact_book:
            print(" Contact book is empty.")
        else:
            print(f"\nTotal contacts: {len(contact_book)}")
            print("-" * 30)
            for name, details in contact_book.items():
                print(f"\n {name}")
                print(f"   Phone: {details['phone']}")
                print(f"   Email: {details['email']}")
    
    # Option 5: Exit
    elif choice == "5":
        print("\n Thank you for using Contact Book!")
        print("Goodbye!")
        break
    
    # Invalid choice
    else:
        print(" Invalid choice! Please enter 1, 2, 3, 4, or 5.")

# Final summary when exiting
print("\n" + "=" * 50)
print("FINAL SUMMARY")
print("=" * 50)
print(f"Total contacts saved: {len(contact_book)}")
if contact_book:
    print("\nContacts:")
    for name in contact_book:
        print(f"   - {name}")
```
